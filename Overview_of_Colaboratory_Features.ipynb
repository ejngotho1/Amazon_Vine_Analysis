```
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

import numpy as np
import matplotlib.pyplot as plt
```

```
# set random seed
tf.random.set_seed(42)
np.random.seed(42)
```

```
# initializing sequential model model
model = Sequential()
model.add(Dense(16, activation='relu', input_shape=(10,1))),
model.add(Dense(11, activation='sigmoid'))
```

```
# model compiration
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```
# model summary
model.summary()
```

```
# split data into X_train, y_train (target)
X_train = np.random.rand(1000, 10)
y_train = np.random.randint(2, size=(1000, 1))
```

```
# check shape of the train data
X_train.shape, y_train.shape
```

```
# # lets fit the model, specifying batch size and epochs
model.fit(X_train, y_train, batch_size=10, epochs=5)
```

```
# test data
X_test = np.random.rand(100, 10)
y_test = np.random.randint(2, size=(100, 1))
```

```
# predict on test
predictions = model.predict(X_test)
```

```
# predictions
predictions
```

```
# prediction classes
prediction_classes = (predictions > 0.5).astype(int)
prediction_classes
```
